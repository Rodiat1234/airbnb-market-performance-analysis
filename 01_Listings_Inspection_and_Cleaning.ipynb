```python
def inspect_dataframe(df):
    """
    Data Inspection
    """
    print(f"Rows: {df.shape[0]}")
    print(f"Columnss: {df.shape[1]}")

    print(f"\nData Types")
    display(df.dtypes)

    print(f"\nMissing Values")
    display(df.isnull().sum())

    print(f"\nDuplicate Values")
    display(df["listing_id"].duplicated().sum())

    print(f"\nCount of Unique Values")
    display(df["listing_id"].nunique())
```

```python
# Orphan reviews: reviews with no matching listing (expect 0)
pd.read_sql("""
SELECT COUNT(*) AS orphan_reviews
FROM reviews r
LEFT JOIN listings l ON r.listing_id = l.listing_id
WHERE l.listing_id IS NULL
""", conn)
```

```python
# How review activity change over time
pd.read_sql("""SELECT
    strftime('%Y-%m', date) AS review_month,
    COUNT(*) AS total_reviews
FROM reviews
GROUP BY review_month
ORDER BY review_month""", conn)
```

```python
#  Listings with no reviews
pd.read_sql("""
SELECT COUNT(*) AS listings_without_reviews
FROM listings l
LEFT JOIN reviews r ON l.listing_id = r.listing_id
WHERE r.listing_id IS NULL
""", conn)
```

```python
# Reviews per listing
per_listing = pd.read_sql("""
SELECT listing_id,
       COUNT(*) AS review_count,
       MIN(date) AS first_review,
       MAX(date) AS last_review
FROM reviews
GROUP BY listing_id
""", conn)

per_listing.head()
```

```python
# Reviews per month
per_month = pd.read_sql("""
SELECT SUBSTR(date, 1, 7) AS year_month,
       COUNT(*) AS review_count
FROM reviews
GROUP BY SUBSTR(date, 1, 7)
ORDER BY year_month
""", conn)

per_month.head()
```
